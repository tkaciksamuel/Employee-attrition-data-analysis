## Phase 3: Attrition study

### Overall attrition rates

```python
attrition_counts = df['Attrition'].value_counts()

print(attrition_counts)
```

```python
attrition_percentages = (
    df['Attrition']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

print(attrition_percentages)
```

```python
plt.figure(figsize=(6, 5))

sns.barplot(
    x=attrition_counts.index,
    y=attrition_counts.values,
)

for i, value in enumerate(attrition_counts.values):
    plt.text(
        i,
        value + 10,
        str(value),
        ha='center'
    )

plt.title('Overall Employee Attrition')
plt.xlabel('Attrition')
plt.ylabel('Number of Employees')

plt.show()
```

```python
retention_rate = (
    (df['Attrition'] == 'No').mean() * 100
).round(2)

print(f"Retention rate: {retention_rate}%")
```

##### Observation: We can notice generally low attrition rates on provided sample with IBM retention rate of 83.88%, further we will study causes of employee parting ways with company and find solutions to make attrition even lower.

### Attrition by department

```python
attrition_by_department = pd.crosstab(
    df['Department'],
    df['Attrition'],
    normalize='index'
) * 100
```

```python
sorted_attrition_by_department = attrition_by_department.sort_values(by='Yes',ascending=False)
```

```python
sorted_attrition_by_department
```

```python
sorted_attrition_by_department = sorted_attrition_by_department[["Yes", "No"]]

sorted_attrition_by_department.plot(
    kind='bar',
    stacked=True,
    figsize=(8,5)
)
plt.title('Employee attrition by department')
plt.xlabel('Department')
plt.ylabel('Attrition')
plt.xticks(rotation=0)
plt.show()
```

##### Observation: From visual representation we can observe that attrition rates among departments are fairly evenly distributed.

### Attrition by marital status 

```python
attrition_counts_marital = pd.crosstab(
    df['MaritalStatus'],
    df['Attrition']
)
```

```python
attrition_correlations = (
    correlation_df
    .corr()['AttritionBinary']
    .drop('AttritionBinary')
    .sort_values(
        key=lambda values: values.abs(),
        ascending=False
    )
    .round(2)
)

attrition_correlations
```

```python
attrition_encoded_df = pd.get_dummies(
    df.drop(
        columns=[
            'Attrition',
            'EmployeeNumber',
            'IncomeGroup',
            'TenureGroup'
        ],
        errors='ignore'
    ),
    drop_first=True,
    dtype=int
)

attrition_encoded_df['AttritionBinary'] = df['Attrition'].map({
    'No': 0,
    'Yes': 1
})
```

### Numerical element correlations

```python
numerical_features = df.select_dtypes(include='number')

numerical_correlation_matrix = numerical_features.corr().round(2)
```

```python
upper_triangle = numerical_correlation_matrix.where(
    np.triu(
        np.ones(numerical_correlation_matrix.shape),
        k=1
    ).astype(bool)
)

strong_numerical_correlations = (
    upper_triangle
    .stack()
    .reset_index()
)

strong_numerical_correlations.columns = [
    'Variate 1',
    'Variate 2',
    'Correlation'
]

strong_numerical_correlations = (
    strong_numerical_correlations[
        strong_numerical_correlations['Correlation'].abs() >= 0.5
    ]
    .sort_values(
        by='Correlation',
        key=lambda values: values.abs(),
        ascending=False
    )
    .reset_index(drop=True)
)

strong_numerical_correlations
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    strong_numerical_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

plt.title(
    'Correlations Between Strongly Related Numerical Features',
    fontsize=14,
    pad=15
)

plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

##### Observation: Based on data provided we can see strongest correlation amongst Job level and Monthly income variates, which suggests that higher position brings higher salary in IBM which is typical. Next we can observe strong corelations with working years and another time related variates with age job level and monthly income ,which suggests that longer period of time spent at company brings higher salary and higher position in firm. Last strong observable correlation is with performance rating and performance salary hike ,which indicates that better performance rating is also cause for salary increase at this company. These are all natural within company position progression ,no particularly strong factors for employee attrition were observed here.

```python
overtime_jobrole = (
    df.groupby(['OverTime', 'JobRole'])['Attrition']
    .apply(lambda x: (x == 'Yes').mean() * 100)
    .reset_index(name='AttritionRate')
)

overtime_jobrole['AttritionRate'] = (
    overtime_jobrole['AttritionRate'].round(2)
)
```

```python
jobrole_overtime_no = jobrole_overtime_no.sort_values(
    by='AttritionRate',
    ascending=False
)

jobrole_overtime_yes = jobrole_overtime_yes.sort_values(
    by='AttritionRate',
    ascending=False
)
```

```python
jobrole_overtime_no = jobrole_overtime_no.reset_index(drop=True)
jobrole_overtime_yes = jobrole_overtime_yes.reset_index(drop=True)
```
