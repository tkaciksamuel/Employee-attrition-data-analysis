### 4.6 Non-Informative Columns

```python
constant_columns = [
    "EmployeeCount",
    "Over18",
    "StandardHours"
]

for col in constant_columns:
    print(f"--- {col} ---")
    print(df[col].value_counts())
    print()
```

**Observation:** `EmployeeCount`, `Over18`, and `StandardHours` contain only one unique value each. Because these columns do not distinguish between employees or provide useful analytical information, they will be removed from the dataset.

```python
df = df.drop(columns=constant_columns)
```

## 5. Derived Columns

### 5.1 Age Groups

```python
age_bins = [17, 25, 35, 45, 55, 60]

age_labels = [
    "18–25",
    "26–35",
    "36–45",
    "46–55",
    "56–60"
]

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels
)
```

### 5.2 Company Tenure Groups

```python
tenure_bins = [0, 2, 5, 10, 20, 40]

tenure_labels = [
    "New Employees (0–2)",
    "Early Career (3–5)",
    "Established (6–10)",
    "Experienced (11–20)",
    "Long Tenure (21+)"
]

df["TenureGroup"] = pd.cut(
    df["YearsAtCompany"],
    bins=tenure_bins,
    labels=tenure_labels,
    include_lowest=True
)
```

### 5.3 Years in Current Role Groups

```python
role_bins = [0, 3, 6, 11, float("inf")]

role_labels = [
    "0–2 years",
    "3–5 years",
    "6–10 years",
    "11+ years"
]

df["YearsInCurrentRoleGroup"] = pd.cut(
    df["YearsInCurrentRole"],
    bins=role_bins,
    labels=role_labels,
    right=False
)
```

### 5.4 Monthly Income Groups

```python
income_bins = [0, 3000, 5000, 8000, 12000, 16000, float("inf")]

income_labels = [
    "Under 3000",
    "3000–4999",
    "5000–7999",
    "8000–11999",
    "12000–15999",
    "16000+"
]

df["IncomeGroup"] = pd.cut(
    df["MonthlyIncome"],
    bins=income_bins,
    labels=income_labels,
    right=False
)
```
