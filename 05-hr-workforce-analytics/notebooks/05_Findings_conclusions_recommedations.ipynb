# IBM HR Attrition — Conclusions and Recommendations

This notebook summarizes the main findings from the workforce, attrition, and multivariate analyses. It highlights the most important employee-turnover patterns and translates them into practical business recommendations.

## 1. Workforce Summary

The dataset contains 1,470 IBM employees across three departments and nine job roles. Research & Development is the largest department, followed by Sales, while Human Resources is the smallest with 63 employees.

The average employee is approximately 37 years old. Sales Executive and Research Scientist are the most common job roles, while the workforce includes employees with varied educational backgrounds, marital statuses, and levels of company tenure.

Most employees travel for business only occasionally, and 71.70% do not work overtime. Monthly income generally increases with job level, reflecting the expected relationship between organizational seniority and employee compensation.

## 2. Attrition Summary

The overall attrition rate within the sampled workforce is 16.12%, although employee turnover is not distributed evenly across the organization.

Sales Representatives show the highest attrition rate among the job roles, while Laboratory Technicians and Human Resources employees also experience elevated turnover. Attrition is more concentrated among younger employees, those with lower job levels, lower monthly incomes, and shorter company tenure.

Overtime emerges as one of the clearest factors associated with employee turnover. Single employees also show a notably high attrition rate of 25.53%, although marital status should be treated as a descriptive workforce characteristic rather than a direct cause of attrition.

## 3. Major Insights

Employee attrition appears to result from a combination of factors rather than one isolated cause. Work-demand characteristics—particularly overtime and frequent business travel—show the clearest relationships with employee turnover.

The highest attrition rates are generally found where demanding working conditions overlap with lower career stability, including younger age, less work experience, lower job level, lower income, and shorter company tenure. Overtime remains associated with increased attrition across departments, job roles, satisfaction levels, travel frequencies, and marital-status groups.

## 4. Business Recommendations

IBM should prioritize reducing excessive overtime, particularly within Sales and other job roles showing elevated employee turnover. Workload distribution, staffing levels, and overtime frequency should be reviewed to identify teams where additional support may be required.

Employees combining overtime with frequent business travel could benefit from more flexible schedules, sufficient recovery time, and better coordination of travel responsibilities. IBM should continue supporting early-career employees by building on its existing career progression structure through effective onboarding, mentoring, development opportunities, and clear communication of available advancement pathways.

Age and marital status should not be used in hiring or employment decisions. These characteristics should only help identify where inclusive employee support and benefits may be improved.

Finally, IBM should monitor attrition regularly and combine workforce data with employee feedback, exit interviews, and engagement surveys to better understand the reasons behind the observed patterns.
